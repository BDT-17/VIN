# CityPersons Dataset Statistics

```python
from pathlib import Path
import json
from collections import defaultdict

DATASET_ROOT = Path("/kaggle/input/cityperson")
IMAGE_ROOT = DATASET_ROOT
LABEL_ROOT = DATASET_ROOT

splits = {"train": DATASET_ROOT / "train", "val": DATASET_ROOT / "val", "test": DATASET_ROOT / "test"}
```

```python
# Count images per split
image_counts = {}
for split, path in splits.items():
    img_dir = path / "images"
    if img_dir.exists():
        image_counts[split] = len(list(img_dir.glob("*.png")))

print("Images per split:")
for split, count in image_counts.items():
    print(f"  {split}: {count}")
print(f"  Total: {sum(image_counts.values())}")
```

```python
# Count annotations per split
annotation_counts = {}
total_annotations = 0

for split, path in splits.items():
    gt_dir = path / "gt"
    if gt_dir.exists():
        anno_files = list(gt_dir.glob("*.txt"))
        annotation_counts[split] = len(anno_files)
        
        # Count total annotations
        split_annos = 0
        for anno_file in anno_files:
            with open(anno_file, 'r') as f:
                split_annos += len(f.readlines())
        total_annotations += split_annos
        print(f"{split}: {annotation_counts[split]} files, {split_annos} annotations")

print(f"\nTotal annotations: {total_annotations}")
```

```python
# Statistics summary
print("\n=== DATASET STATISTICS ===")
print(f"Total images: {sum(image_counts.values())}")
print(f"Total annotations: {total_annotations}")
print(f"Avg annotations per image: {total_annotations / sum(image_counts.values()):.2f}")
print(f"\nSplit breakdown:")
for split in splits.keys():
    print(f"  {split}: {image_counts.get(split, 0)} images")
```
